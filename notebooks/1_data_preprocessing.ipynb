```python
# ==================================================
# CELL 1: Import Library & Setup
# ==================================================
# Tujuan: Menyiapkan environment dan library yang dibutuhkan

import pandas as pd
import numpy as np
import os

# Menampilkan semua kolom agar mudah dicek
pd.set_option('display.max_columns', None)

print("Environment siap! Library berhasil diimpor.")
```

```python
# ==================================================
# CELL 2: Load Data Mentah (Raw Data)
# ==================================================
# Tujuan: Membaca dataset mentah dari folder data/raw

# Sesuaikan path jika nama file sedikit berbeda
df_master = pd.read_csv('../data/raw/players_data-2024_2025.csv')  

print(f"Dimensi data mentah: {df_master.shape}")
print("\n--- Daftar 20 Kolom Pertama ---")
print(df_master.columns[:20].tolist())

display(df_master.head())
```

```python
# ==================================================
# CELL 3: Merge Data + Seleksi Fitur + Cleaning + Log Transform
# ==================================================
# Tujuan:
# 1. Gabungkan statistik pemain dengan nilai pasar
# 2. Pilih fitur yang digunakan
# 3. Cleaning (duplikat + missing)
# 4. Transformasi logaritma

# 1. Siapkan data nilai pasar dari players.csv
df_players = pd.read_csv('../data/raw/players.csv')

# Ambil kolom yang dibutuhkan saja
df_nilai = df_players[['name', 'market_value_in_eur']].copy()
df_nilai = df_nilai.dropna(subset=['market_value_in_eur'])
df_nilai = df_nilai.rename(columns={'name': 'Player'})

print(f"Jumlah pemain yang punya nilai pasar: {df_nilai.shape[0]}")

# 2. Gabungkan dengan data statistik (df_master)
df_merged = df_master.merge(df_nilai, on='Player', how='inner')

print(f"Jumlah data setelah merge: {df_merged.shape[0]}")

# 3. Seleksi fitur (Skenario B / Fitur Lengkap)
kolom_skenario_b = [
    'Player', 'Age', 'Pos', 'Comp',
    'Gls', 'Ast',
    'xG', 'xA', 'SCA', 'PrgP',
    'market_value_in_eur'
]

df_model = df_merged[kolom_skenario_b].copy()

# 4. Data Cleaning
df_model = df_model.drop_duplicates(subset=['Player'], keep='first')
df_model = df_model.dropna()

# Opsional: buang penjaga gawang jika ingin fokus outfield
# df_model = df_model[~df_model['Pos'].str.contains('GK', na=False)]

# 5. Transformasi Logaritma
df_model['Log_Market_Value'] = np.log1p(df_model['market_value_in_eur'])

print("\n--- HASIL PRA-PEMROSESAN ---")
print(f"Dimensi data final: {df_model.shape}")
display(df_model[['Player', 'Pos', 'Comp', 'market_value_in_eur', 'Log_Market_Value']].head())
```

```python
# ==================================================
# CELL 4: Filter Pemain Outfield + Liga Top
# ==================================================
# Tujuan: Menyelaraskan data dengan batasan penelitian

print("Jumlah data sebelum filter:", df_model.shape[0])

# 1. Buang Penjaga Gawang (GK)
df_model = df_model[~df_model['Pos'].astype(str).str.contains('GK', na=False)]

# 2. (Opsional) Hanya 5 Liga Top Eropa
liga_top = [
    'eng Premier League',
    'es La Liga', 
    'it Serie A',
    'de Bundesliga',
    'fr Ligue 1'
]
# Uncomment baris di bawah jika ingin filter liga
# df_model = df_model[df_model['Comp'].isin(liga_top)]

print("Jumlah data setelah filter GK:", df_model.shape[0])
display(df_model['Pos'].value_counts())
```

```python
# ==================================================
# CELL 5: One-Hot Encoding & Simpan Data Siap Model
# ==================================================
# Tujuan: Mengubah data kategorikal menjadi numerik dan menyimpan hasil akhir

print("Memulai tahap akhir pra-pemrosesan...")

# One-Hot Encoding (drop_first=True untuk hindari dummy variable trap)
df_final = pd.get_dummies(df_model, columns=['Pos', 'Comp'], drop_first=True)

# Simpan nama pemain (jika dibutuhkan nanti) lalu buang kolom yang tidak dipakai model
df_nama = df_final[['Player']].copy()
df_final = df_final.drop(columns=['Player', 'market_value_in_eur'])

# Pastikan folder processed ada
os.makedirs('../data/processed', exist_ok=True)

# Simpan data final
df_final.to_csv('../data/processed/data_siap_model_skenario_b.csv', index=False)

print("\n--- STATUS AKHIR ---")
print(f"Dimensi data final setelah Encoding: {df_final.shape}")
print("SUKSES! Data disimpan ke: '../data/processed/data_siap_model_skenario_b.csv'")
display(df_final.head())
```
